# Plotting from pandas

```python
import pandas as pd
```

```python
car_sales = pd.read_csv("/Users/kamari/Downloads/car-sales.csv")
car_sales
```

```python
car_sales["Price"]= car_sales["Price"].str.replace('[\$\,\.]', '', regex=True)
car_sales
```

```python
car_sales["Price"] = car_sales["Price"].str[:-2]
car_sales
```

```python
ex_1 = car_sales["Make"].value_counts()
ex_1.plot(kind="bar")
plt.title("Car Make Distribution")
plt.show()  
```

```python
car_sales["Odometer (KM)"].plot.hist(title="Odometer Distribution",bins=10)
plt.xlabel("Odometer (KM)")
plt.ylabel("Frequency")
plt.show() 
```

```python
x = car_sales["Odometer (KM)"]
y = car_sales["Price"]
fig, ax = plt.subplots()
ax.scatter(x, y)
plt.xlabel("Odometer (KM)")
plt.ylabel("Price")
plt.show()  

```

```python
fig, ax = plt.subplots(1, 2, figsize=(10,5))
car_sales["Odometer (KM)"].plot.hist(title="Odometer Distribution", bins=10, ax=ax[0])
car_sales.plot.scatter(x="Odometer (KM)", 
                       y="Price", 
                       title="Price vs Odometer", 
                       ax=ax[1])
plt.tight_layout()
plt.show()
```

```python
car_sales["Price"] = car_sales["Price"].astype(float)
```

```python
car_sales["Price"].mean()
```

```python
car_sales.groupby("Make")["Price"].mean().plot(kind="bar")
plt.title("Average Price by Car Make")
plt.show()
```

```python
car_sales["Colour"].value_counts().plot(kind="bar")
plt.title("Car Colour Distribution")
plt.xlabel("Colour")
plt.ylabel("Frequency") 
plt.show()
```

```python
car_sales.groupby("Doors")["Price"].mean().plot(kind="bar")
plt.title("Average Price by Number of Doors")
plt.xlabel("Number of Doors")
plt.ylabel("Average Price")
plt.show()
```

```python
car_sales.groupby("Colour")["Price"].mean().sort_values(ascending=False).plot(kind="bar")
plt.show()
```

```python
car_sales
```

```python
car_sales.groupby("Make").agg({"Price": ["mean"], 
                               "Odometer (KM)": ["mean"], 
                               "Make": ["count"]
                               })
```

```python
fig, ax = plt.subplots(2, 2, figsize=(12,8))
car_sales.groupby("Make")["Price"].mean().plot(kind="bar", ax=ax[0,0])

car_sales["Colour"].value_counts().plot(kind="bar", ax=ax[0,1])

car_sales["Odometer (KM)"].plot(kind="hist", ax=ax[1,0])

car_sales.plot(kind="scatter",
               x = "Odometer (KM)",
               y = "Price",
               ax=ax[1,1])

plt.tight_layout()
plt.show()
```
