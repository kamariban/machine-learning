```python
#digit counter
def count_dig(item):
    counter = 0
    digits = str(item)
    for x in digits:
        if "0"<= x <="9":
            counter+= 1
    return counter


def count_upper_letter(item):
    counter = 0
    letters = str(item)
    for x in letters:
        if "A"<= x <="Z":
            counter +=1
    return  counter
```

```python

def reverse_vowels(word):
    vowels = ["a","e","i","o","u"]
    store = []

    new_word = str(word)
    for x in new_word:
        if x in vowels:
            store.append(x)
    #reverse
    store.reverse()

    # Replace vowels one at a time
    result = ""
    position = 0

    for x in new_word:
        if x in vowels:
            result += store[position]
            position += 1
        else:
            result += x

    return result
print(reverse_vowels("haelloup"))
```

```python
#move digits to the end
def move_digits(word):
    nums = []
    letters = []
    new_word = str(word)
    for x in new_word:
        if "0"<= x <="9":
            nums.append(x)
        else:
            letters.append(x)
    final_str = ''.join(letters)
    final_num = ''.join(nums)
    return final_str + final_num

print(move_digits("12312-3--4hello"))
```

```python
#return duplicates
def first_dupe(li):
    dupes = []
    for x in list(li):
        if x not in dupes:
            dupes.append(x)
        else:
            return x

    return -1
print(first_dupe([1,2,3,34,3,2,2,]))
```

```python

def string_compress(word):
    counter = []
    compressed = []
    for x in word:
        if x not in compressed:
            counter.append(x)
            chr_count = str(word.count(x))
            counter.append(chr_count)
            compressed.append(x)
    fin = "".join(counter)
    return fin
```

```python

#most frequent character
def most_frequency(word):
    char_dict = {}
    count = 0
    ans = ''
    for x in str(word):
        char_dict[x] = str(word).count(x)
    for k, v in char_dict.items():
        if v > count:
            count = v
            ans = k
    return ans
#print(most_frequency("Halloooo"))
```

```python

#missing num
def missing_number(li):
    order = []
    missing = []
    for x in range(1,max(li)+1):
        order.append(x)
    for y in order:
        if y not in li:
            missing.append(y)
    return missing

print(missing_number([5, 1, 2, 3]))
```

```python

#word scores
def word_score(sent):
    letter_dict = {}
    character_code = 97
    letter_score = 1
    my_word_score = 0
    for x in range(1, 27):
        letter_dict[chr(character_code)] = letter_score
        character_code+= 1
        letter_score += 1
    for x in sent.split():
        for y in x:
            my_word_score +=  letter_dict[y]
    return my_word_score
print(word_score("dog d cat"))
```

```python
def first_dupe(li):
    exists = []
    for x in li:
        if x not in exists:
            exists.append(x)
        elif x in exists:
            return x
    return -1
print(first_dupe([1,2,3]))
```

```python

#2. the gap number
def missing_num(li):
    new_list = sorted(li)
    counter = new_list[0]
    for x in new_list:
        if x == counter:
            counter += 1
        else:
            return counter
    return counter

print(missing_num([3,1,0]))
```

```python
#2.5 unique number
def unique_num(li):
    scores = {}
    new_list = sorted(li)
    for x in new_list:
        scores[x] = new_list.count(x)
    for k, v in scores.items():
        if v == 1:
            return k
    return scores

#print(unique_num([7, 3, 5, 3, 7]))
```

```python

#5 move zeros
def move_zeros(li):
    moved_zeros = []
    counter = 0
    for x in li:
        if x != 0:
            moved_zeros.append(x)
        if x == 0:
            counter += 1
    for y in range(counter):
        moved_zeros.append(0)

    return moved_zeros

print(move_zeros([10, 0, 1, 0, 3, 12]))
```

```python
#6 Majority Element

def majority(li):
    gutter = {}
    for x in sorted(li):
        gutter[x] = li.count(x)
        if li.count(x) > (len(li) / 2):
            return x
print(majority([2, 2, 3, ]))
```
